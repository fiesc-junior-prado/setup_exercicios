# Exercício I

import pandas as pd
import numpy as np

# Definindo a semente para garantir que os dados gerados sejam os mesmos sempre
np.random.seed(42)

# Gerando os dados simulados
# Investimento mensal em marketing varia entre 1k e 15k
investimento = np.linspace(1, 15, 50) 

# As vendas têm uma base de 20k + 4.5k para cada 1k investido + um ruído aleatório do mundo real
vendas = 20 + 4.5 * investimento + np.random.normal(0, 5, 50) 

# Criando o DataFrame
df_cafeteria = pd.DataFrame({
    'Investimento_Marketing_Mil_RS': investimento,
    'Vendas_Mil_RS': vendas
})

print(df_cafeteria.head())

# Exercício II

import pandas as pd
import numpy as np

# Definindo a semente para garantir a reprodutibilidade
np.random.seed(42)
n_amostras = 500

# Gerando as variáveis independentes (Features)
distancia = np.random.uniform(5, 100, n_amostras)      # Distância em km
peso = np.random.uniform(0.5, 30, n_amostras)          # Peso em kg
transito = np.random.randint(1, 11, n_amostras)        # Índice de trânsito (1 a 10)
chuva = np.random.uniform(0, 50, n_amostras)           # Volume de chuva em mm

# Simulando o tempo de entrega com relações não-lineares e interações
# Ex: Chuva afeta muito mais quando o trânsito já está ruim
tempo_entrega = (
    (distancia * 0.15) + 
    (peso * 0.05) + 
    (transito ** 1.3 * 0.25) + 
    (chuva * 0.03 * transito) + 
    np.random.normal(0, 2, n_amostras) # Ruído aleatório da vida real
)

# Garantindo que não exista tempo de entrega negativo ou irreal
tempo_entrega = np.maximum(1.0, tempo_entrega)

# Criando o DataFrame
df_logistica = pd.DataFrame({
    'Distancia_km': distancia,
    'Peso_kg': peso,
    'Transito_Indice': transito,
    'Chuva_mm': chuva,
    'Tempo_Entrega_Horas': tempo_entrega
})

print(df_logistica.head())

# Exercício_III

import pandas as pd
import numpy as np

# Definindo a semente para garantir reprodutibilidade
np.random.seed(42)

# Simulando 4 perfis de comportamento ocultos (100 clientes em cada)
# Grupo A: Acessam muito o app, mas compram pouco (Exploradores de vitrine)
g1_acessos = np.random.normal(40, 5, 100)
g1_gastos = np.random.normal(150, 40, 100)

# Grupo B: Acessam pouco e compram pouco (Compradores ocasionais)
g2_acessos = np.random.normal(10, 3, 100)
g2_gastos = np.random.normal(100, 30, 100)

# Grupo C: Acessam muito e gastam muito (Clientes VIP / Fãs da marca)
g3_acessos = np.random.normal(35, 6, 100)
g3_gastos = np.random.normal(800, 120, 100)

# Grupo D: Acessam pouco, mas gastam muito (Compradores diretos/decididos)
g4_acessos = np.random.normal(8, 2, 100)
g4_gastos = np.random.normal(750, 100, 100)

# Juntando e embaralhando os dados para simular um banco de dados real
acessos = np.concatenate([g1_acessos, g2_acessos, g3_acessos, g4_acessos])
gastos = np.concatenate([g1_gastos, g2_gastos, g3_gastos, g4_gastos])

df_clientes = pd.DataFrame({
    'Acessos_App_Mes': np.maximum(1, np.round(acessos).astype(int)), # Evita acessos negativos
    'Gasto_Mensal_RS': np.maximum(10, np.round(gastos, 2))           # Evita gastos negativos
})

# Embaralhando as linhas
df_clientes = df_clientes.sample(frac=1, random_state=42).reset_index(drop=True)

print("Amostra dos dados extraídos:")
print(df_clientes.head())
