## Exercicio I

```python
import pandas as pd
from sklearn.datasets import make_blobs

def gerar_dataset_impressao_3d():
    """
    Gera um dataset sintético para classificação binária linearmente separável.
    Representa testes de impressão 3D com PETG.
    """
    # Gerando 2 clusters distintos (Sucesso vs Falha)
    # Centro 1 (Falha): Temp mais baixa (~230C), Vel mais alta (~70 mm/s)
    # Centro 2 (Sucesso): Temp mais alta (~250C), Vel mais controlada (~45 mm/s)
    X, y = make_blobs(
        n_samples=200, 
        centers=[[230, 70], [250, 45]], 
        cluster_std=4.0, # Mantém os dados aglomerados mas com uma margem clara
        random_state=42
    )
    
    # Montando o DataFrame
    df = pd.DataFrame(X, columns=['Temperatura_Bico_C', 'Velocidade_mms'])
    df['Status_Impressao'] = y # 0: Falha, 1: Sucesso
    
    # Arredondando para simular dados reais de fatiador
    df['Temperatura_Bico_C'] = df['Temperatura_Bico_C'].round(1)
    df['Velocidade_mms'] = df['Velocidade_mms'].round(1)
    
    return df

# Executando a geração
df_testes = gerar_dataset_impressao_3d()

# Exibindo as 5 primeiras linhas para conferência
print("Amostra do Dataset Gerado:")
print(df_testes.head())

# Opcional: Salvar para enviar aos alunos
# df_testes.to_csv('testes_petg_impressao.csv', index=False)
```

## Exercicio II

```python
import numpy as np
import pandas as pd

def gerar_dataset_corte_laser_mdf():
    """
    Gera um dataset sintético para classificação binária de corte a laser em MDF de 3mm.
    """
    np.random.seed(42)
    n_amostras = 150
    
    # Gerando velocidades aleatórias entre 100 e 500 mm/min
    velocidade = np.random.uniform(100, 500, n_amostras)
    
    # Gerando potências aleatórias entre 40% e 100%
    potencia = np.random.uniform(40, 100, n_amostras)
    
    status_corte = []
    
    for v, p in zip(velocidade, potencia):
        # Regra linear física hipotética: 
        # Para cortar, a potência precisa ser maior que uma proporção da velocidade.
        # Ex: Para 300 mm/min, precisa de pelo menos 70% de potência (300 * 0.15 + 25 = 70)
        limite_potencia = (v * 0.15) + 25 
        
        # Adicionando um pequeno ruído para simular variações na densidade do MDF
        ruido = np.random.normal(0, 3) 
        
        if p >= (limite_potencia + ruido):
            status_corte.append(1) # Sucesso (Atravessou o MDF)
        else:
            status_corte.append(0) # Falha (Não cortou)
            
    df = pd.DataFrame({
        'Velocidade_mm_min': np.round(velocidade, 0),
        'Potencia_Percentual': np.round(potencia, 1),
        'Status_Corte': status_corte
    })
    
    return df

# Executando a geração
df_laser = gerar_dataset_corte_laser_mdf()

# Exibindo a distribuição
print("Distribuição das classes:")
print(df_laser['Status_Corte'].value_counts())
print("\nAmostra do Dataset:")
print(df_laser.head())

# df_laser.to_csv('parametros_corte_mdf.csv', index=False)
```

## Exercício III

```python
import pandas as pd
from sklearn.datasets import make_circles

def gerar_dataset_motores_iot():
    """
    Gera um dataset sintético não-linear (círculos concêntricos) 
    para simular dados de sensores IoT de manutenção preditiva.
    """
    # n_samples: quantidade de leituras
    # noise: o nível de ruído/mistura entre as classes
    # factor: a distância entre o círculo interno e o externo
    X, y = make_circles(n_samples=300, noise=0.1, factor=0.5, random_state=42)
    
    # Os dados do make_circles vêm em uma escala de -1 a 1. 
    # Vamos converter para valores que façam sentido físico:
    
    # Vibração variando de ~30Hz a ~70Hz (Centro em 50Hz)
    vibracao = (X[:, 0] * 20) + 50
    
    # Temperatura variando de ~30°C a ~80°C (Centro em 55°C)
    temperatura = (X[:, 1] * 25) + 55
    
    df = pd.DataFrame({
        'Vibracao_Hz': round(pd.Series(vibracao), 2),
        'Temperatura_C': round(pd.Series(temperatura), 2),
        'Status': y  # 0: Normal (Círculo interno), 1: Anomalia (Círculo externo)
    })
    
    return df

# Executando a geração
df_motores = gerar_dataset_motores_iot()

print("Amostra do Dataset Gerado:")
print(df_motores.head())
print("\nContagem de Classes:")
print(df_motores['Status'].value_counts())

# df_motores.to_csv('sensores_motores_iot.csv', index=False)
```

## Exercício IV

```python
import pandas as pd
from sklearn.datasets import make_moons

def gerar_dataset_gestos_smartwatch():
    """
    Gera um dataset sintético não-linear (meias-luas entrelaçadas) 
    para simular dados de um acelerômetro reconhecendo gestos.
    """
    # noise: simula a imperfeição do movimento humano e imprecisão do sensor
    X, y = make_moons(n_samples=500, noise=0.15, random_state=42)
    
    # Dimensionando os dados para parecerem leituras de um acelerômetro (m/s²)
    # Multiplicamos por 10 e adicionamos um leve deslocamento
    aceleracao_x = X[:, 0] * 10
    aceleracao_y = X[:, 1] * 10
    
    df = pd.DataFrame({
        'Aceleracao_X_ms2': round(pd.Series(aceleracao_x), 3),
        'Aceleracao_Y_ms2': round(pd.Series(aceleracao_y), 3),
        'Gesto': y  # 0: Deslize Lateral, 1: Giro de Pulso
    })
    
    return df

# Executando a geração
df_gestos = gerar_dataset_gestos_smartwatch()

print("Amostra do Dataset Gerado:")
print(df_gestos.head())
print("\nContagem de Gestos:")
print(df_gestos['Gesto'].value_counts())

# Opcional: Salvar para enviar aos alunos
# df_gestos.to_csv('gestos_smartwatch.csv', index=False)
```
