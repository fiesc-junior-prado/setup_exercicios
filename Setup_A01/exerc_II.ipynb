```python
import pandas as pd
import numpy as np

# Fixando a semente para garantir os mesmos dados
np.random.seed(101)
n_amostras = 1200

# Gerando o perfil dos funcionários
idade = np.random.randint(20, 60, size=n_amostras)

# Salário com base na idade, mas com variações
salario = (idade * 150) + np.random.normal(2000, 1500, size=n_amostras)
salario = np.clip(salario, 2000, 20000) # Salário mínimo de 2k e máximo de 20k

# Pesquisa de clima (1 a 10) e controle de ponto (horas extras no mês)
satisfacao = np.random.randint(1, 11, size=n_amostras)
horas_extras = np.random.randint(0, 50, size=n_amostras)

# A regra de negócio oculta que o modelo precisará descobrir:
# - Insatisfação e excesso de horas extras aumentam drasticamente a chance de sair.
# - Salários mais altos e maior idade reduzem a chance de sair.
formula_rh = 4.5 - (0.05 * idade) - (0.0003 * salario) - (0.8 * satisfacao) + (0.15 * horas_extras)

# Transformando em probabilidade com a função Sigmoide
probabilidade = 1 / (1 + np.exp(-formula_rh))

# Se a probabilidade for > 50%, o funcionário pediu demissão
vai_sair = (probabilidade > 0.5).astype(int)

# Montando o DataFrame
df_rh = pd.DataFrame({
    'ID_Funcionario': np.arange(5001, 5001 + n_amostras),
    'Idade': idade,
    'Salario_Reais': np.round(salario, 2),
    'Nivel_Satisfacao': satisfacao,
    'Horas_Extras_Mes': horas_extras,
    'Vai_Sair': vai_sair
})

print(f"Base do RH gerada com sucesso! Total de registros: {df_rh.shape[0]}")
print(df_rh.head())
```
