```python
import pandas as pd
import numpy as np

# Fixando a semente para garantir que os dados gerados sejam sempre os mesmos
np.random.seed(42)
n_amostras = 1000

# Gerando as características (features) com alguma correlação lógica
idades = np.random.randint(18, 70, size=n_amostras)
anos_emprego = np.maximum(0, idades - 18 - np.random.randint(0, 5, size=n_amostras))
renda_mensal = (idades * 50) + (anos_emprego * 200) + np.random.normal(1000, 500, size=n_amostras)

# Score Serasa é influenciado pela renda
score_serasa = np.clip(300 + (renda_mensal / 20) + np.random.normal(0, 80, size=n_amostras), 300, 1000)

# Criando a regra oculta de aprovação (Regressão Logística perfeita para isso)
# Maior renda e maior score aumentam a chance de aprovação
formula_oculta = -12 + (score_serasa * 0.01) + (renda_mensal * 0.001)
probabilidade = 1 / (1 + np.exp(-formula_oculta))
aprovado = (probabilidade > 0.5).astype(int)

# Montando o DataFrame final
df_credito = pd.DataFrame({
    'ID_Cliente': np.arange(1001, 1001 + n_amostras),
    'Idade': idades,
    'Renda_Mensal': np.round(renda_mensal, 2),
    'Anos_Emprego_Atual': anos_emprego,
    'Score_Serasa': np.round(score_serasa).astype(int),
    'Aprovado': aprovado
})

print(f"Base gerada com sucesso! Total de registros: {df_credito.shape[0]}")
print(df_credito.head())
```
